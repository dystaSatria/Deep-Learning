```python
# Gerekli kütüphaneleri içe aktarma
from sklearn.ensemble import AdaBoostClassifier
from sklearn.tree import DecisionTreeClassifier
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import accuracy_score
import numpy as np

# Örnek veri seti oluşturma
X, y = make_classification(n_samples=1000, n_features=4, n_classes=2, random_state=42)

# Veriyi eğitim ve test setlerine ayırma
X_train, X_test, y_train, y_test = train_test_split(X, y, test_size=0.3, random_state=42)

# AdaBoost modeli oluşturma
ada_model = AdaBoostClassifier(
    estimator=DecisionTreeClassifier(max_depth=1),  # Zayıf öğrenici (decision stump)
    n_estimators=50,  # 50 zayıf öğrenici
    learning_rate=1.0,  # Öğrenme oranı
    random_state=42
)

# Modeli eğitme
ada_model.fit(X_train, y_train)

# Tahmin yapma
y_pred = ada_model.predict(X_test)

# Sonuçları yazdırma
print(f"AdaBoost Doğruluk Oranı: {accuracy_score(y_test, y_pred):.4f}")
print(f"Kullanılan Estimator Sayısı: {len(ada_model.estimators_)}")
```

```python
# Gradient Hesaplama Örnekleri
import numpy as np
import matplotlib.pyplot as plt
from mpl_toolkits.mplot3d import Axes3D
import sympy as sp

# ==========================================
# ÖRNEK 1: TEK DEĞİŞKENLİ FONKSİYON TÜREVİ
# ==========================================

print("=== TEK DEĞİŞKENLİ FONKSİYON TÜREVİ ===")

# f(x) = x^2 + 3x + 2 fonksiyonu
def f_single(x):
    return x**2 + 3*x + 2

# Türevi: f'(x) = 2x + 3
def f_prime(x):
    return 2*x + 3

# Sayısal türev hesaplama
def numerical_derivative(f, x, h=1e-5):
    return (f(x + h) - f(x - h)) / (2 * h)

# Test
x_test = 2
analytical = f_prime(x_test)
numerical = numerical_derivative(f_single, x_test)

print(f"x = {x_test} noktasında:")
print(f"Analitik türev: {analytical}")
print(f"Sayısal türev: {numerical:.5f}")

# ==========================================
# ÖRNEK 2: İKİ DEĞİŞKENLİ FONKSİYON GRADİENTİ
# ==========================================

print("\n=== İKİ DEĞİŞKENLİ FONKSİYON GRADİENTİ ===")

# f(x,y) = x^2 + y^2 (basit paraboloid)
def f_2d(x, y):
    return x**2 + y**2

# Gradient: ∇f = (∂f/∂x, ∂f/∂y) = (2x, 2y)
def gradient_2d(x, y):
    df_dx = 2 * x  # ∂f/∂x
    df_dy = 2 * y  # ∂f/∂y
    return np.array([df_dx, df_dy])

# Sayısal gradient hesaplama
def numerical_gradient_2d(f, x, y, h=1e-5):
    df_dx = (f(x + h, y) - f(x - h, y)) / (2 * h)
    df_dy = (f(x, y + h) - f(x, y - h)) / (2 * h)
    return np.array([df_dx, df_dy])

# Test
x_test, y_test = 3, 4
grad_analytical = gradient_2d(x_test, y_test)
grad_numerical = numerical_gradient_2d(f_2d, x_test, y_test)

print(f"({x_test}, {y_test}) noktasında:")
print(f"Analitik gradient: {grad_analytical}")
print(f"Sayısal gradient: {grad_numerical}")
print(f"Gradient büyüklüğü: {np.linalg.norm(grad_analytical):.2f}")

# ==========================================
# ÖRNEK 3: KARMAŞIK FONKSİYON GRADİENTİ
# ==========================================

print("\n=== KARMAŞIK FONKSİYON GRADİENTİ ===")

# f(x,y) = x^2*y + y^3 - 2*x*y
def f_complex(x, y):
    return x**2 * y + y**3 - 2*x*y

# Gradient hesaplama
# ∂f/∂x = 2xy - 2y = 2y(x-1)
# ∂f/∂y = x^2 + 3y^2 - 2x
def gradient_complex(x, y):
    df_dx = 2*y*(x - 1)
    df_dy = x**2 + 3*y**2 - 2*x
    return np.array([df_dx, df_dy])

# SymPy ile otomatik türev
x_sym, y_sym = sp.symbols('x y')
f_sym = x_sym**2 * y_sym + y_sym**3 - 2*x_sym*y_sym

# Kısmi türevler
df_dx_sym = sp.diff(f_sym, x_sym)
df_dy_sym = sp.diff(f_sym, y_sym)

print("SymPy ile hesaplanan türevler:")
print(f"∂f/∂x = {df_dx_sym}")
print(f"∂f/∂y = {df_dy_sym}")

# Belirli bir noktada değerlendirme
x_val, y_val = 2, 1
grad_manual = gradient_complex(x_val, y_val)
grad_sympy = np.array([
    float(df_dx_sym.subs([(x_sym, x_val), (y_sym, y_val)])),
    float(df_dy_sym.subs([(x_sym, x_val), (y_sym, y_val)]))
])

print(f"\n({x_val}, {y_val}) noktasında:")
print(f"Manuel hesaplama: {grad_manual}")
print(f"SymPy hesaplama: {grad_sympy}")

# ==========================================
# ÖRNEK 4: GRADIENT DESCENT ALGORİTMASI
# ==========================================

print("\n=== GRADIENT DESCENT ALGORİTMASI ===")

def gradient_descent(f, grad_f, start_point, learning_rate=0.1, max_iter=100, tolerance=1e-6):
    """Basit gradient descent algoritması"""
    point = np.array(start_point, dtype=float)
    path = [point.copy()]

    for i in range(max_iter):
        gradient = grad_f(point[0], point[1])
        new_point = point - learning_rate * gradient

        # Yakınsama kontrolü
        if np.linalg.norm(new_point - point) < tolerance:
            print(f"Yakınsama {i+1}. iterasyonda sağlandı")
            break

        point = new_point
        path.append(point.copy())

    return point, np.array(path)

# f(x,y) = x^2 + y^2 fonksiyonu için minimum bulma
start = [5, 3]  # Başlangıç noktası
minimum, path = gradient_descent(f_2d, gradient_2d, start, learning_rate=0.1)

print(f"Başlangıç noktası: {start}")
print(f"Bulunan minimum: {minimum}")
print(f"Minimum'daki fonksiyon değeri: {f_2d(minimum[0], minimum[1]):.6f}")
print(f"Toplam iterasyon: {len(path)}")

# ==========================================
# ÖRNEK 5: ÜÇLÜ GRADIENT (3D)
# ==========================================

print("\n=== ÜÇLÜ GRADIENT (3D) ===")

# f(x,y,z) = x^2 + y^2 + z^2
def f_3d(x, y, z):
    return x**2 + y**2 + z**2

# Gradient: ∇f = (2x, 2y, 2z)
def gradient_3d(x, y, z):
    return np.array([2*x, 2*y, 2*z])

# Test
x, y, z = 1, 2, 3
grad_3d = gradient_3d(x, y, z)
print(f"({x}, {y}, {z}) noktasında gradient: {grad_3d}")
print(f"Gradient büyüklüğü: {np.linalg.norm(grad_3d):.2f}")

# ==========================================
# ÖRNEK 6: SICAKLIK GRADİENTİ SİMÜLASYONU
# ==========================================

print("\n=== SICAKLIK GRADİENTİ SİMÜLASYONU ===")

def temperature_field(x, y):
    """2D sıcaklık alanı simülasyonu"""
    # Merkezi ısı kaynağı
    center_x, center_y = 0, 0
    distance = np.sqrt((x - center_x)**2 + (y - center_y)**2)
    return 100 * np.exp(-distance**2 / 4)  # Gaussian ısı dağılımı

def temperature_gradient(x, y):
    """Sıcaklık gradientini hesaplama"""
    h = 1e-5
    dx = (temperature_field(x + h, y) - temperature_field(x - h, y)) / (2 * h)
    dy = (temperature_field(x, y + h) - temperature_field(x, y - h)) / (2 * h)
    return np.array([dx, dy])

# Farklı noktalarda sıcaklık gradientini hesaplama
test_points = [(1, 0), (0, 1), (1, 1), (2, 2)]

print("Sıcaklık alanında gradient analizi:")
for point in test_points:
    x, y = point
    temp = temperature_field(x, y)
    grad = temperature_gradient(x, y)
    grad_magnitude = np.linalg.norm(grad)

    print(f"Nokta ({x}, {y}):")
    print(f"  Sıcaklık: {temp:.2f}°C")
    print(f"  Gradient: ({grad[0]:.2f}, {grad[1]:.2f})")
    print(f"  Gradient büyüklüğü: {grad_magnitude:.2f}")
    print()

# ==========================================
# ÖRNEK 7: VİZUALİZASYON İÇİN KOD HAZIRLIĞI
# ==========================================

print("=== VİZUALİZASYON VERİSİ HAZIRLIĞI ===")

# 2D grid oluşturma
x_range = np.linspace(-3, 3, 20)
y_range = np.linspace(-3, 3, 20)
X, Y = np.meshgrid(x_range, y_range)

# Fonksiyon değerleri
Z = f_2d(X, Y)

# Gradient bileşenleri
U, V = np.zeros_like(X), np.zeros_like(Y)
for i in range(len(x_range)):
    for j in range(len(y_range)):
        grad = gradient_2d(X[i,j], Y[i,j])
        U[i,j] = grad[0]
        V[i,j] = grad[1]

print("Gradient field verileri hazırlandı.")
print(f"Grid boyutu: {X.shape}")
print("Bu veriler matplotlib ile görselleştirilebilir:")
print("- Kontur haritası: plt.contour(X, Y, Z)")
print("- Gradient vektörleri: plt.quiver(X, Y, U, V)")

# ==========================================
# BONUS: PRATIK GRADIENT HESAP MAKİNESİ
# ==========================================

print("\n=== PRATIK GRADIENT HESAP MAKİNESİ ===")

class GradientCalculator:
    def __init__(self):
        self.h = 1e-5  # Sayısal türev için adım büyüklüğü

    def calculate_1d(self, func, x):
        """Tek değişkenli fonksiyon türevi"""
        return (func(x + self.h) - func(x - self.h)) / (2 * self.h)

    def calculate_2d(self, func, x, y):
        """İki değişkenli fonksiyon gradientı"""
        dx = (func(x + self.h, y) - func(x - self.h, y)) / (2 * self.h)
        dy = (func(x, y + self.h) - func(x, y - self.h)) / (2 * self.h)
        return np.array([dx, dy])

    def calculate_3d(self, func, x, y, z):
        """Üç değişkenli fonksiyon gradientı"""
        dx = (func(x + self.h, y, z) - func(x - self.h, y, z)) / (2 * self.h)
        dy = (func(x, y + self.h, z) - func(x, y - self.h, z)) / (2 * self.h)
        dz = (func(x, y, z + self.h) - func(x, y, z - self.h)) / (2 * self.h)
        return np.array([dx, dy, dz])

# Hesap makinesi kullanımı
calc = GradientCalculator()

# Test fonksiyonları
def test_func_1d(x):
    return x**3 - 2*x**2 + x

def test_func_2d(x, y):
    return x**2 + x*y + y**2

def test_func_3d(x, y, z):
    return x**2 + y**2 + z**2 + x*y*z

# Hesaplamalar
print("Gradient Hesap Makinesi Test Sonuçları:")
print(f"f(x) = x³-2x²+x, x=2: f'(2) = {calc.calculate_1d(test_func_1d, 2):.3f}")
print(f"f(x,y) = x²+xy+y², (1,1): ∇f = {calc.calculate_2d(test_func_2d, 1, 1)}")
print(f"f(x,y,z) = x²+y²+z²+xyz, (1,1,1): ∇f = {calc.calculate_3d(test_func_3d, 1, 1, 1)}")

print("\n=== KULLANIM KILAVUZU ===")
print("1. Tek değişken: calc.calculate_1d(fonksiyon, x_değeri)")
print("2. İki değişken: calc.calculate_2d(fonksiyon, x_değeri, y_değeri)")
print("3. Üç değişken: calc.calculate_3d(fonksiyon, x_değeri, y_değeri, z_değeri)")
print("4. Gradient büyüklüğü: np.linalg.norm(gradient_vektörü)")
print("5. Gradient yönü: gradient_vektörü / np.linalg.norm(gradient_vektörü)")
```
