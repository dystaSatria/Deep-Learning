```python
import tensorflow as tf
from tensorflow.keras import layers, Model
import numpy as np
import matplotlib.pyplot as plt
from sklearn.metrics import classification_report, confusion_matrix
import seaborn as sns

# Verdiğiniz DenseNet kodunu burada kullanacağız
# (DenseBlock, TransitionLayer, DenseNet sınıflarınız burada tanımlı olmalı)

# 1. Basit Model Oluşturma ve Test Etme
def create_simple_model_example():
    """
    Basit bir DenseNet modeli oluşturma örneği
    """
    print("=== Basit DenseNet Modeli Oluşturma ===")

    try:
        # Küçük bir DenseNet modeli oluştur
        model = DenseNet121(num_classes=10)  # CIFAR-10 için 10 sınıf

        # Modeli build etmek için dummy input kullan
        dummy_input = tf.random.normal((1, 32, 32, 3))
        _ = model(dummy_input)

        # Model yapısını göster
        print("\nModel yapısı:")
        model.summary()

        # Model parametrelerini göster
        total_params = model.count_params()
        print(f"\nToplam parametre sayısı: {total_params:,}")

        return model

    except Exception as e:
        print(f"Model oluşturma hatası: {e}")
        print("Alternatif küçük model oluşturuluyor...")

        # Daha basit alternatif model
        model = DenseNet([2, 4], growth_rate=12, num_classes=10)
        dummy_input = tf.random.normal((1, 32, 32, 3))
        _ = model(dummy_input)

        total_params = model.count_params()
        print(f"Alternatif model parametreleri: {total_params:,}")

        return model

# 2. CIFAR-10 ile Hızlı Eğitim Örneği
def quick_training_example():
    """
    CIFAR-10 veri seti ile hızlı eğitim örneği
    """
    print("\n=== Hızlı Eğitim Örneği ===")

    # Veri setini yükle
    (x_train, y_train), (x_test, y_test) = tf.keras.datasets.cifar10.load_data()

    # Küçük bir subset kullan (hızlı test için)
    x_train_small = x_train[:1000]
    y_train_small = y_train[:1000]
    x_test_small = x_test[:200]
    y_test_small = y_test[:200]

    # Veri ön işleme
    x_train_small = x_train_small.astype('float32') / 255.0
    x_test_small = x_test_small.astype('float32') / 255.0

    # One-hot encoding
    y_train_small = tf.keras.utils.to_categorical(y_train_small, 10)
    y_test_small = tf.keras.utils.to_categorical(y_test_small, 10)

    # Küçük bir model oluştur
    model = DenseNet([2, 4, 6, 4], growth_rate=16, num_classes=10)

    # Model derle
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.01),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    # Kısa eğitim
    print("Model eğitimi başlıyor...")
    history = model.fit(
        x_train_small, y_train_small,
        batch_size=32,
        epochs=5,
        validation_data=(x_test_small, y_test_small),
        verbose=1
    )

    # Sonuçları değerlendir
    test_loss, test_acc = model.evaluate(x_test_small, y_test_small, verbose=0)
    print(f"\nTest doğruluğu: {test_acc:.4f}")

    return model, history

# 3. Transfer Learning Örneği
def transfer_learning_example():
    """
    Transfer learning ile önceden eğitilmiş model kullanma
    """
    print("\n=== Transfer Learning Örneği ===")

    # Önceden eğitilmiş DenseNet121 modeli yükle
    base_model = tf.keras.applications.DenseNet121(
        weights='imagenet',
        include_top=False,
        input_shape=(224, 224, 3)
    )

    # Feature extraction için katmanları dondur
    base_model.trainable = False

    # Yeni sınıflandırıcı ekle
    model = tf.keras.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dropout(0.3),
        layers.Dense(256, activation='relu'),
        layers.Dropout(0.3),
        layers.Dense(5, activation='softmax')  # 5 sınıf için
    ])

    # Model derle
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
        loss='categorical_crossentropy',
        metrics=['accuracy']
    )

    print("Transfer learning modeli hazır!")
    model.summary()

    return model

# 4. Özellik Haritası Görselleştirme Örneği
def feature_visualization_example():
    """
    DenseNet katmanlarından özellik haritalarını görselleştirme
    """
    print("\n=== Özellik Haritası Görselleştirme ===")

    try:
        # Küçük bir model oluştur
        model = DenseNet([2, 4], growth_rate=8, num_classes=10)

        # Rastgele bir test görüntüsü oluştur
        test_image = np.random.rand(1, 32, 32, 3).astype('float32')

        # Modeli ilk defa çağır (build için)
        _ = model(test_image)

        # İlk konvolüsyon katmanının çıktısını al
        intermediate_model = Model(
            inputs=model.input,
            outputs=model.conv1.output
        )

        # Özellik haritalarını çıkar
        feature_maps = intermediate_model(test_image)

        # Görselleştir
        num_maps = min(8, feature_maps.shape[-1])
        fig, axes = plt.subplots(2, 4, figsize=(12, 6))
        axes = axes.ravel()

        for i in range(num_maps):
            if i < len(axes):
                axes[i].imshow(feature_maps[0, :, :, i], cmap='viridis')
                axes[i].set_title(f'Feature Map {i+1}')
                axes[i].axis('off')

        # Kullanılmayan subplot'ları gizle
        for i in range(num_maps, len(axes)):
            axes[i].axis('off')

        plt.suptitle('DenseNet İlk Katman Özellik Haritaları')
        plt.tight_layout()
        plt.show()

        print(f"Özellik haritaları başarıyla görselleştirildi: {feature_maps.shape}")

        return feature_maps

    except Exception as e:
        print(f"Özellik görselleştirme hatası: {e}")
        print("Alternatif basit görselleştirme yapılıyor...")

        # Basit alternatif
        test_data = np.random.rand(2, 4, 32, 32)
        fig, axes = plt.subplots(2, 4, figsize=(12, 6))
        axes = axes.ravel()

        for i in range(8):
            axes[i].imshow(test_data[i//4, i%4], cmap='viridis')
            axes[i].set_title(f'Sample Map {i+1}')
            axes[i].axis('off')

        plt.suptitle('Örnek Özellik Haritaları (Simulated)')
        plt.tight_layout()
        plt.show()

        return test_data

# 5. Model Karşılaştırma Örneği
def model_comparison_example():
    """
    Farklı DenseNet varyantlarını karşılaştırma
    """
    print("\n=== Model Karşılaştırma ===")

    models_info = []

    # Farklı DenseNet konfigürasyonları
    configs = [
        ("DenseNet-Small", [2, 4, 6, 4], 16),
        ("DenseNet-Medium", [4, 8, 12, 8], 24),
        ("DenseNet-Large", [6, 12, 24, 16], 32)
    ]

    for name, blocks, growth_rate in configs:
        try:
            model = DenseNet(blocks, growth_rate=growth_rate, num_classes=10)

            # Modeli build etmek için dummy input kullan
            dummy_input = tf.random.normal((1, 32, 32, 3))
            _ = model(dummy_input)

            param_count = model.count_params()
            models_info.append({
                'Model': name,
                'Bloklar': str(blocks),
                'Growth Rate': growth_rate,
                'Parametreler': f"{param_count:,}"
            })

            print(f"{name}: {param_count:,} parametre")

        except Exception as e:
            print(f"{name} oluşturulamadı: {e}")
            models_info.append({
                'Model': name,
                'Bloklar': str(blocks),
                'Growth Rate': growth_rate,
                'Parametreler': "Hata"
            })

    return models_info

# 6. Custom Loss Function ile Eğitim
def custom_training_example():
    """
    Özel loss function ve metriklerle eğitim
    """
    print("\n=== Özel Loss Function ile Eğitim ===")

    # Özel loss function tanımla
    def focal_loss(gamma=2., alpha=0.25):
        def focal_loss_fixed(y_true, y_pred):
            epsilon = tf.keras.backend.epsilon()
            y_pred = tf.clip_by_value(y_pred, epsilon, 1. - epsilon)

            p_t = tf.where(tf.equal(y_true, 1), y_pred, 1 - y_pred)
            alpha_factor = tf.ones_like(y_true) * alpha
            alpha_t = tf.where(tf.equal(y_true, 1), alpha_factor, 1 - alpha_factor)
            cross_entropy = -tf.math.log(p_t)
            weight = alpha_t * tf.pow((1 - p_t), gamma)

            loss = weight * cross_entropy
            return tf.reduce_mean(tf.reduce_sum(loss, axis=1))

        return focal_loss_fixed

    # Model oluştur
    model = DenseNet([2, 4], growth_rate=12, num_classes=10)

    # Özel loss ile derle
    model.compile(
        optimizer=tf.keras.optimizers.Adam(learning_rate=0.001),
        loss=focal_loss(gamma=2.0, alpha=0.25),
        metrics=['accuracy']
    )

    print("Özel focal loss function ile model derlendi!")

    return model

# 7. Gradual Unfreezing Örneği
def gradual_unfreezing_example():
    """
    Transfer learning'de aşamalı katman çözme
    """
    print("\n=== Gradual Unfreezing Örneği ===")

    # Önceden eğitilmiş model yükle
    base_model = tf.keras.applications.DenseNet121(
        weights='imagenet',
        include_top=False,
        input_shape=(224, 224, 3)
    )

    # Tüm katmanları dondur
    base_model.trainable = False

    # Model oluştur
    model = tf.keras.Sequential([
        base_model,
        layers.GlobalAveragePooling2D(),
        layers.Dense(128, activation='relu'),
        layers.Dense(10, activation='softmax')
    ])

    def unfreeze_layers(model, num_layers):
        """Belirtilen sayıda katmanı çöz"""
        base_model = model.layers[0]

        # Tüm katmanları dondur
        base_model.trainable = False

        # Son num_layers katmanını çöz
        for layer in base_model.layers[-num_layers:]:
            layer.trainable = True

        # Düşük learning rate ile yeniden derle
        model.compile(
            optimizer=tf.keras.optimizers.Adam(learning_rate=0.0001),
            loss='categorical_crossentropy',
            metrics=['accuracy']
        )

        print(f"{num_layers} katman çözüldü")

    # Aşamalı çözme stratejisi
    unfreeze_strategy = [0, 10, 25, 50]

    for step, num_layers in enumerate(unfreeze_strategy):
        print(f"Adım {step + 1}: {num_layers} katman çözülüyor")
        unfreeze_layers(model, num_layers)

    return model

# 8. Model Ensemble Örneği
def ensemble_example():
    """
    Birden fazla DenseNet modelini ensemble etme
    """
    print("\n=== Model Ensemble Örneği ===")

    # Farklı konfigürasyonlarda modeller oluştur
    models = []

    configs = [
        ([2, 4, 6, 4], 16),
        ([3, 6, 8, 6], 20),
        ([4, 8, 10, 8], 24)
    ]

    for i, (blocks, growth_rate) in enumerate(configs):
        try:
            model = DenseNet(blocks, growth_rate=growth_rate, num_classes=10)

            # Modeli build etmek için dummy input kullan
            dummy_input = tf.random.normal((1, 32, 32, 3))
            _ = model(dummy_input)

            models.append(model)
            print(f"Model {i+1} oluşturuldu: blocks={blocks}, growth_rate={growth_rate}")

        except Exception as e:
            print(f"Model {i+1} oluşturulamadı: {e}")

    if len(models) == 0:
        print("Hiçbir model oluşturulamadı, basit alternatif kullanılıyor...")
        # Basit alternatif modeller
        for i in range(3):
            model = DenseNet([2, 4], growth_rate=8, num_classes=10)
            dummy_input = tf.random.normal((1, 32, 32, 3))
            _ = model(dummy_input)
            models.append(model)

    # Ensemble prediction fonksiyonu
    def ensemble_predict(models, x):
        if len(models) == 0:
            return None

        predictions = []
        for model in models:
            try:
                pred = model(x)
                predictions.append(pred)
            except Exception as e:
                print(f"Model prediction hatası: {e}")
                continue

        if len(predictions) == 0:
            return None

        # Ortalama al
        ensemble_pred = tf.reduce_mean(predictions, axis=0)
        return ensemble_pred

    print(f"{len(models)} model ensemble'ı hazır!")

    return models, ensemble_predict

# Test ve Debug Fonksiyonu
def test_densenet_basic():
    """
    DenseNet'in temel fonksiyonalitesini test eder
    """
    print("=== DenseNet Temel Test ===")

    try:
        # En basit DenseNet oluştur
        print("1. Basit DenseNet oluşturuluyor...")
        model = DenseNet([2, 2], growth_rate=8, num_classes=5)

        # Test verisi oluştur
        print("2. Test verisi hazırlanıyor...")
        test_input = tf.random.normal((2, 32, 32, 3))

        # Model çağrısı
        print("3. Model çağrılıyor...")
        output = model(test_input)

        print(f"Başarılı! Input shape: {test_input.shape}, Output shape: {output.shape}")
        print(f"Model parametreleri: {model.count_params():,}")

        return True, model

    except Exception as e:
        print(f"Test başarısız: {e}")
        return False, None

# Ana çalıştırma fonksiyonu
def run_all_examples():
    """
    Tüm örnekleri sırayla çalıştır
    """
    print("DenseNet Pratik Örnekleri Başlıyor...\n")

    # Önce temel test yap
    success, test_model = test_densenet_basic()
    if not success:
        print("Temel test başarısız, sadece basit örnekler çalıştırılacak...")
        return {'test': 'failed'}

    results = {'test_model': test_model}

    try:
        # 1. Basit model oluşturma
        print("\n" + "="*50)
        model1 = create_simple_model_example()
        results['simple_model'] = model1
    except Exception as e:
        print(f"Basit model örneği hatası: {e}")

    try:
        # 2. Hızlı eğitim
        print("\n" + "="*50)
        model2, history = quick_training_example()
        results['trained_model'] = model2
        results['training_history'] = history
    except Exception as e:
        print(f"Hızlı eğitim örneği hatası: {e}")

    try:
        # 3. Transfer learning
        print("\n" + "="*50)
        model3 = transfer_learning_example()
        results['transfer_model'] = model3
    except Exception as e:
        print(f"Transfer learning örneği hatası: {e}")

    try:
        # 4. Özellik görselleştirme
        print("\n" + "="*50)
        features = feature_visualization_example()
        results['features'] = features
    except Exception as e:
        print(f"Özellik görselleştirme hatası: {e}")

    try:
        # 5. Model karşılaştırma
        print("\n" + "="*50)
        comparison = model_comparison_example()
        results['comparison'] = comparison
    except Exception as e:
        print(f"Model karşılaştırma hatası: {e}")

    try:
        # 6. Özel loss function
        print("\n" + "="*50)
        model6 = custom_training_example()
        results['custom_loss_model'] = model6
    except Exception as e:
        print(f"Özel loss function hatası: {e}")

    try:
        # 7. Gradual unfreezing
        print("\n" + "="*50)
        model7 = gradual_unfreezing_example()
        results['unfreezing_model'] = model7
    except Exception as e:
        print(f"Gradual unfreezing hatası: {e}")

    try:
        # 8. Ensemble
        print("\n" + "="*50)
        ensemble_models, ensemble_func = ensemble_example()
        results['ensemble_models'] = ensemble_models
        results['ensemble_function'] = ensemble_func
    except Exception as e:
        print(f"Ensemble örneği hatası: {e}")

    print("\n" + "="*50)
    print("=== Örnekler Tamamlandı! ===")
    print(f"Başarılı olan örnekler: {len([k for k in results.keys() if k != 'test_model'])}")

    return results

# Kullanım örneği
if __name__ == "__main__":
    # Tüm örnekleri çalıştır
    results = run_all_examples()

    # Eğitim geçmişini görselleştir
    if 'training_history' in results and results['training_history']:
        history = results['training_history']

        plt.figure(figsize=(12, 4))

        plt.subplot(1, 2, 1)
        plt.plot(history.history['loss'], label='Eğitim Kaybı')
        plt.plot(history.history['val_loss'], label='Doğrulama Kaybı')
        plt.title('Model Kaybı')
        plt.xlabel('Epoch')
        plt.ylabel('Kayıp')
        plt.legend()

        plt.subplot(1, 2, 2)
        plt.plot(history.history['accuracy'], label='Eğitim Doğruluğu')
        plt.plot(history.history['val_accuracy'], label='Doğrulama Doğruluğu')
        plt.title('Model Doğruluğu')
        plt.xlabel('Epoch')
        plt.ylabel('Doğruluk')
        plt.legend()

        plt.tight_layout()
        plt.show()

    print("Örnekler başarıyla tamamlandı!")
```
