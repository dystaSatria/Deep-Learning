```python
import numpy as np
import matplotlib.pyplot as plt

print("=== LINEAR ALGEBRA BASICS UNTUK DEEP LEARNING ===\n")

# 1. MATRIKS BASICS
print("1. MATRIKS BASICS")
print("-" * 30)

# Membuat matriks (seperti tabel Excel)
A = np.array([[1, 2, 3],
              [4, 5, 6]])  # Matriks 2x3

B = np.array([[7, 8],
              [9, 10],
              [11, 12]])   # Matriks 3x2

print("Matriks A (2x3):")
print(A)
print(f"Shape: {A.shape}")

print("\nMatriks B (3x2):")
print(B)
print(f"Shape: {B.shape}")

# 2. DOT PRODUCT (MATRIX MULTIPLICATION)
print("\n2. DOT PRODUCT (Matrix Multiplication)")
print("-" * 45)

# Ini operasi paling penting dalam neural networks!
C = np.dot(A, B)  # atau A @ B
print("A × B =")
print(C)
print(f"Shape hasil: {C.shape}")

# Contoh sederhana: 1 neuron dengan 3 input
print("\n--- Contoh: 1 Neuron dengan 3 Input ---")
inputs = np.array([0.5, 0.3, 0.8])      # Input data
weights = np.array([0.2, 0.7, 0.1])     # Weights neuron
bias = 0.1                               # Bias

# Operasi neuron: weighted sum + bias
output = np.dot(inputs, weights) + bias
print(f"Inputs: {inputs}")
print(f"Weights: {weights}")
print(f"Output: {output:.3f}")

# 3. MATRIX OPERATIONS UNTUK MULTIPLE SAMPLES
print("\n3. BATCH PROCESSING (Multiple Samples)")
print("-" * 40)

# Simulasi 4 samples, masing-masing 3 features
X = np.array([[1, 2, 3],      # Sample 1
              [4, 5, 6],      # Sample 2
              [7, 8, 9],      # Sample 3
              [2, 1, 4]])     # Sample 4

# Weights untuk 2 neurons
W = np.array([[0.1, 0.2],    # Weights neuron 1 & 2 untuk feature 1
              [0.3, 0.4],    # Weights neuron 1 & 2 untuk feature 2
              [0.5, 0.6]])   # Weights neuron 1 & 2 untuk feature 3

print("Input data X (4 samples, 3 features):")
print(X)
print(f"Shape: {X.shape}")

print("\nWeights W (3 features, 2 neurons):")
print(W)
print(f"Shape: {W.shape}")

# Forward pass untuk semua samples sekaligus
outputs = np.dot(X, W)
print("\nOutputs (4 samples, 2 neurons):")
print(outputs)
print(f"Shape: {outputs.shape}")

# 4. ELEMENT-WISE OPERATIONS
print("\n4. ELEMENT-WISE OPERATIONS")
print("-" * 30)

# Operations yang diterapkan ke setiap elemen
matrix1 = np.array([[1, 2], [3, 4]])
matrix2 = np.array([[5, 6], [7, 8]])

print("Matrix 1:")
print(matrix1)
print("Matrix 2:")
print(matrix2)

print("\nAddition (element-wise):")
print(matrix1 + matrix2)

print("\nMultiplication (element-wise):")
print(matrix1 * matrix2)

print("\nSquare (element-wise):")
print(matrix1 ** 2)

# 5. TRANSFORMASI LINEAR - CONTOH REAL
print("\n5. TRANSFORMASI LINEAR - Contoh Rotasi 2D")
print("-" * 45)

# Data points (x, y coordinates)
points = np.array([[1, 0],     # Point 1
                   [0, 1],     # Point 2
                   [1, 1],     # Point 3
                   [2, 1]])    # Point 4

# Rotation matrix (45 degrees)
angle = np.pi / 4  # 45 degrees in radians
rotation_matrix = np.array([[np.cos(angle), -np.sin(angle)],
                           [np.sin(angle),  np.cos(angle)]])

print("Original points:")
print(points)

print("\nRotation matrix (45°):")
print(rotation_matrix)

# Apply transformation
rotated_points = np.dot(points, rotation_matrix.T)
print("\nRotated points:")
print(rotated_points)

# Visualisasi
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.scatter(points[:, 0], points[:, 1], c='blue', s=100, label='Original')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.title('Original Points')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

plt.subplot(1, 2, 2)
plt.scatter(rotated_points[:, 0], rotated_points[:, 1], c='red', s=100, label='Rotated')
plt.grid(True, alpha=0.3)
plt.axis('equal')
plt.title('After 45° Rotation')
plt.xlabel('X')
plt.ylabel('Y')
plt.legend()

plt.tight_layout()
plt.savefig('linear_transformation.png', dpi=150, bbox_inches='tight')
plt.show()

# 6. CONTOH MINI NEURAL NETWORK LAYER
print("\n6. MINI NEURAL NETWORK LAYER")
print("-" * 35)

def linear_layer(X, W, b):
    """
    Implementasi sederhana linear layer
    X: input data (samples x features)
    W: weights (features x neurons)
    b: bias (neurons)
    """
    return np.dot(X, W) + b

def relu_activation(x):
    """ReLU activation function"""
    return np.maximum(0, x)

# Simulasi data
np.random.seed(42)
X_input = np.random.randn(5, 3)  # 5 samples, 3 features
W1 = np.random.randn(3, 4) * 0.5 # 3 features -> 4 neurons
b1 = np.zeros(4)                 # bias untuk 4 neurons

print("Input data shape:", X_input.shape)
print("Weights shape:", W1.shape)
print("Bias shape:", b1.shape)

# Forward pass
linear_output = linear_layer(X_input, W1, b1)
activated_output = relu_activation(linear_output)

print("\nLinear output (before activation):")
print(linear_output[:3])  # Show first 3 samples

print("\nAfter ReLU activation:")
print(activated_output[:3])  # Show first 3 samples

print("\n=== KESIMPULAN ===")
print("✓ Matriks adalah struktur data dasar dalam neural networks")
print("✓ Dot product digunakan untuk menghitung weighted sum")
print("✓ Transformasi linear mengubah data dari satu space ke space lain")
print("✓ Batch processing memungkinkan training yang efisien")
print("✓ Semua operasi neural network adalah kombinasi linear algebra!")
```
